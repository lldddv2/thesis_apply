```python
from relatipy.numeric.metrics import Kerr
from relatipy.numeric.coordinates import ApparentOrbitalElements
from relatipy.numeric.geodesic import Geodesic
from relatipy.visualization._3D import PlotKerr
```

```python
# --- 1. Métrica de Kerr con spin a/M = 0.9 ---
kerr = Kerr(mass=1.0, a=0.9)

# --- 2. Condiciones iniciales en elementos orbitales aparentes (plano del cielo) ---
ic = ApparentOrbitalElements(
    t=0.0, a=30.0, e=0.5,          # semieje mayor [GM/c²], excentricidad
    inc=45.0, Omega=20.0, omega=10.0, f=0.0,  # ángulos orbitales [°]
    zeta=30.0, eta=0.0,             # orientación del spin del BH respecto al observador [°]
    mass=1.0,
)
```

```python
# --- 3. Periodo orbital kepleriano y grilla de tiempo propio ---
T = ic._get_period()               # Kepler III en unidades geométricas [GM/c³]
taus = [0, 5*T]

# --- 4. Integrar geodésica y convertir a Cartesiano ---
path = Geodesic(kerr).get_path(ic, taus, integrator="Radau2")
```

```python
# --- 5. Visualización 3D ---
plot = PlotKerr(kerr)
plot.add_path(path)
plot.plot()
```
